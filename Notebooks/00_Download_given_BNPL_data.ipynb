****Now Download the 4 given zip files on Canvas for BNPL to the "data/given" folder just created**

```python
import zipfile

def unzip_parts(zip_dir, output_dir):
    """
    Unzip all 'project-2-bnpl-tables-partX.zip' files into folders 'part X'.

    Args:
        zip_dir (str): Directory containing zip files.
        output_dir (str): Directory where extracted folders will be saved.
    """
    os.makedirs(output_dir, exist_ok=True)

    for filename in os.listdir(zip_dir):
        if filename.startswith("project-2-bnpl-tables-part") and filename.endswith(".zip"):
            # Extract part number
            part_number = filename.split("part")[-1].split(".")[0]
            folder_name = f"part {part_number}"
            extract_path = os.path.join(output_dir, folder_name)
            os.makedirs(extract_path, exist_ok=True)

            zip_path = os.path.join(zip_dir, filename)
            with zipfile.ZipFile(zip_path, 'r') as zip_ref:
                zip_ref.extractall(extract_path)

            print(f"Extracted {filename} -> {extract_path}")


# Example usage
zip_directory = "../data/given"      # folder with the .zip files
output_directory = "../data/given"

unzip_parts(zip_directory, output_directory)
```
